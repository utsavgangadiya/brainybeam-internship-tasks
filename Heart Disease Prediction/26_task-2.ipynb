Split the dataset into training (80%) and testing (20%) without using built-in train-test split functions, ensuring class balance and stratification through manual implementation.

```python
import pandas as pd
import random

# Load dataset
df = pd.read_csv("final_cleaned_heart_disease.csv")

target_column = "target"

print("Dataset loaded successfully!")
print("Total rows:", len(df))
print("Target distribution:")
print(df[target_column].value_counts())
```

```python
# Set random seed for reproducible results
random.seed(42)

# Store indices for each target class
class_indices = {}

for i in range(len(df)):

    target_value = df[target_column].iloc[i]

    if target_value not in class_indices:
        class_indices[target_value] = []

    class_indices[target_value].append(i)


# Create training and testing indices
train_indices = []
test_indices = []


# Split each class separately
for target_value in class_indices:

    indices = class_indices[target_value]

    # Shuffle indices manually
    random.shuffle(indices)

    # Calculate 80% training size
    train_size = int(
        len(indices) * 0.80
    )

    # Add 80% to training
    for i in range(train_size):

        train_indices.append(
            indices[i]
        )

    # Add remaining 20% to testing
    for i in range(
        train_size,
        len(indices)
    ):

        test_indices.append(
            indices[i]
        )


# Shuffle final datasets
random.shuffle(train_indices)
random.shuffle(test_indices)


# Create datasets
train_df = df.iloc[
    train_indices
].reset_index(drop=True)

test_df = df.iloc[
    test_indices
].reset_index(drop=True)


print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print(
    "\nTraining percentage:",
    round(
        len(train_df) / len(df) * 100,
        2
    ),
    "%"
)

print(
    "Testing percentage:",
    round(
        len(test_df) / len(df) * 100,
        2
    ),
    "%"
)
```

```python
# Check class distribution
print("\nTraining class distribution:")

train_counts = {}

for value in train_df[target_column]:

    if value not in train_counts:
        train_counts[value] = 0

    train_counts[value] += 1


for value in train_counts:

    percentage = (
        train_counts[value]
        /
        len(train_df)
        *
        100
    )

    print(
        "Class",
        value,
        ":",
        train_counts[value],
        "(",
        round(percentage, 2),
        "%)"
    )


print("\nTesting class distribution:")

test_counts = {}

for value in test_df[target_column]:

    if value not in test_counts:
        test_counts[value] = 0

    test_counts[value] += 1


for value in test_counts:

    percentage = (
        test_counts[value]
        /
        len(test_df)
        *
        100
    )

    print(
        "Class",
        value,
        ":",
        test_counts[value],
        "(",
        round(percentage, 2),
        "%)"
    )
```

```python
# Save training dataset
train_df.to_csv(
    "heart_disease_train.csv",
    index=False
)

# Save testing dataset
test_df.to_csv(
    "heart_disease_test.csv",
    index=False
)

print(
    "\nTraining dataset saved as:",
    "heart_disease_train.csv"
)

print(
    "Testing dataset saved as:",
    "heart_disease_test.csv"
)
```
