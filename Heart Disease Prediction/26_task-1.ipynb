Store the final cleaned dataset in a CSV file by writing a custom function that formats the data, ensures proper encoding, and optimizes file size without using built-in DataFrame export functions.

```python
import pandas as pd
import os

# Load final cleaned dataset
df = pd.read_csv("heart_disease_with_risk_factor.csv")

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
def format_csv_value(value):

    # Handle missing values
    if pd.isna(value):
        return ""

    # Convert numbers efficiently
    if isinstance(value, float):

        if value.is_integer():
            return str(int(value))

        return str(round(value, 6))

    # Convert other values to string
    value = str(value)

    # Handle commas and quotation marks
    if "," in value or '"' in value:

        value = value.replace(
            '"',
            '""'
        )

        return '"' + value + '"'

    return value


def save_csv_manually(dataframe, filename):

    with open(
        filename,
        "w",
        encoding="utf-8",
        newline=""
    ) as file:

        # Write column names
        headers = []

        for column in dataframe.columns:

            headers.append(
                format_csv_value(column)
            )

        file.write(
            ",".join(headers)
            + "\n"
        )

        # Write data rows
        for i in range(len(dataframe)):

            row_values = []

            for column in dataframe.columns:

                value = dataframe[
                    column
                ].iloc[i]

                formatted_value = (
                    format_csv_value(value)
                )

                row_values.append(
                    formatted_value
                )

            file.write(
                ",".join(row_values)
                + "\n"
            )

    print(
        "CSV file created successfully!"
    )

    print(
        "File name:",
        filename
    )


# Save final cleaned dataset
output_file = "final_cleaned_heart_disease.csv"

save_csv_manually(
    df,
    output_file
)
```

```python
# Check file size
file_size = os.path.getsize(
    output_file
)

print(
    "File size:",
    file_size,
    "bytes"
)

print(
    "File size:",
    round(
        file_size / 1024,
        2
    ),
    "KB"
)
```

```python
# Read the generated file for verification
check_df = pd.read_csv(
    output_file
)

print(
    "File verification successful!"
)

print(
    check_df.head()
)

print(
    "\nRows:",
    len(check_df)
)

print(
    "Columns:",
    len(check_df.columns)
)
```
