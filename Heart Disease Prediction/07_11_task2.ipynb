Predicting Stock Market Prices	Engineer a reliable data type transformation pipeline that accurately converts and parses date-related fields into a consistent datetime format suitable for time-indexed operations.

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("TSLA_cleaned_data.csv")

print("Dataset Shape:", data.shape)
display(data.head())
```

```python
print("Date Column Data Type:", data["Date"].dtype)

print("\nSample Date Values:")
display(data["Date"].head(10))
```

```python
data["Date"] = pd.to_datetime(
    data["Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

print("Date Conversion Completed!")
print("New Data Type:", data["Date"].dtype)
```

```python
invalid_dates = data["Date"].isna().sum()

print("Invalid or Missing Dates:", invalid_dates)

if invalid_dates > 0:
    print("\nInvalid Date Records:")
    display(data[data["Date"].isna()])
```

```python
print("Date Data Type:", data["Date"].dtype)

print("Is Datetime:", pd.api.types.is_datetime64_any_dtype(data["Date"]))

print("Earliest Date:", data["Date"].min())
print("Latest Date:", data["Date"].max())
```

```python
clean_data = data.dropna(subset=["Date"]).copy()

print("Original Records:", len(data))
print("Records After Date Cleaning:", len(clean_data))
```

```python
clean_data["Year"] = clean_data["Date"].dt.year
clean_data["Month"] = clean_data["Date"].dt.month
clean_data["Day"] = clean_data["Date"].dt.day
clean_data["Day_Name"] = clean_data["Date"].dt.day_name()
clean_data["Weekday"] = clean_data["Date"].dt.dayofweek

display(clean_data.head())
```

```python
clean_data = clean_data.sort_values("Date")

print("Dates Sorted:", clean_data["Date"].is_monotonic_increasing)

display(clean_data.head())
```

```python
time_series_data = clean_data.set_index("Date")

print("Index Data Type:", time_series_data.index.dtype)
print("Index Name:", time_series_data.index.name)

display(time_series_data.head())
```

```python
date_report = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Invalid Dates",
        "Missing Dates",
        "Date Data Type",
        "Earliest Date",
        "Latest Date",
        "Chronological Order"
    ],
    "Result": [
        len(time_series_data),
        invalid_dates,
        time_series_data.index.isna().sum(),
        str(time_series_data.index.dtype),
        time_series_data.index.min(),
        time_series_data.index.max(),
        time_series_data.index.is_monotonic_increasing
    ]
})

display(date_report)
```

```python
time_series_data.to_csv("TSLA_datetime_processed.csv")

print("Date-transformed dataset saved successfully!")
```
