Test UI for usability by conducting user experience research, gathering qualitative feedback, identifying pain points in interaction flow, and iteratively refining the interface based on usability heuristics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output
```

```python
test_tasks = [
    "Open the Heart Disease Prediction interface",
    "Understand the input fields",
    "Enter patient measurements",
    "Identify and correct invalid inputs",
    "Generate a prediction",
    "Understand the prediction confidence",
    "Interpret the final result"
]

for i, task in enumerate(test_tasks, 1):
    print(f"Task {i}: {task}")
```

```python
participant = widgets.Text(
    description="Tester ID:",
    placeholder="Example: User 1"
)

ease = widgets.IntSlider(
    value=3, min=1, max=5,
    description="Ease of use:"
)

navigation = widgets.IntSlider(
    value=3, min=1, max=5,
    description="Navigation:"
)

clarity = widgets.IntSlider(
    value=3, min=1, max=5,
    description="Result clarity:"
)

satisfaction = widgets.IntSlider(
    value=3, min=1, max=5,
    description="Satisfaction:"
)

feedback = widgets.Textarea(
    description="Feedback:",
    placeholder="What was confusing or difficult?",
    layout=widgets.Layout(width="90%", height="100px")
)

pain_points = widgets.Textarea(
    description="Pain points:",
    placeholder="Describe problems encountered",
    layout=widgets.Layout(width="90%", height="80px")
)

suggestions = widgets.Textarea(
    description="Suggestions:",
    placeholder="What should be improved?",
    layout=widgets.Layout(width="90%", height="80px")
)

display(widgets.VBox([
    participant,
    ease,
    navigation,
    clarity,
    satisfaction,
    feedback,
    pain_points,
    suggestions
]))
```

```python
feedback_data = []

def save_feedback(button):

    if not participant.value.strip():
        print("Please enter a tester ID.")
        return

    feedback_data.append({
        "Tester": participant.value,
        "Ease of Use": ease.value,
        "Navigation": navigation.value,
        "Result Clarity": clarity.value,
        "Satisfaction": satisfaction.value,
        "Feedback": feedback.value,
        "Pain Points": pain_points.value,
        "Suggestions": suggestions.value
    })

    print("Feedback saved successfully!")

save_button = widgets.Button(
    description="Submit Feedback",
    button_style="success"
)

save_button.on_click(save_feedback)

display(save_button)
```

```python
feedback_df = pd.DataFrame(feedback_data)

if not feedback_df.empty:

    display(feedback_df)

    print("\nCommon Pain Points:")

    for point in feedback_df["Pain Points"]:
        if point.strip():
            print("-", point)

else:
    print("No feedback collected yet. Complete user testing first.")
```

```python
heuristics = {
    "Visibility of system status": "Does the interface show loading and prediction status?",
    "Error prevention": "Does it prevent missing or incorrect inputs?",
    "Consistency": "Are labels, buttons and layouts consistent?",
    "User control": "Can users edit inputs and correct mistakes?",
    "Understandable feedback": "Are prediction results clearly explained?"
}

for name, question in heuristics.items():
    print(f"\n{name}:")
    print(question)
```

```python
improvements = {
    "Input fields": "Add clear labels, units and example values.",
    "Input validation": "Show error messages next to incorrect fields.",
    "Prediction button": "Provide clear prediction status.",
    "Prediction result": "Use readable result labels and confidence percentages.",
    "Navigation": "Keep the input and prediction sections organized.",
    "Accessibility": "Use readable text and consistent spacing."
}

for issue, solution in improvements.items():
    print(f"\nIssue: {issue}")
    print(f"Improvement: {solution}")
```

```python
if not feedback_df.empty:

    score_columns = [
        "Ease of Use",
        "Navigation",
        "Result Clarity",
        "Satisfaction"
    ]

    average_scores = feedback_df[score_columns].mean()

    average_scores.plot(
        kind="bar",
        figsize=(8, 4),
        ylim=(0, 5),
        title="Average User Experience Scores"
    )

    plt.ylabel("Average Rating (1–5)")
    plt.xlabel("Usability Factors")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

else:
    print("Collect user feedback before generating the chart.")
```

```python
if not feedback_df.empty:
    feedback_df.to_csv("usability_feedback_report.csv", index=False)
    print("Feedback report exported successfully!")
else:
    print("No feedback available to export.")
```
