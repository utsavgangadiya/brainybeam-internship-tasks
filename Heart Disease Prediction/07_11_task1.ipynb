```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv(
    "TSLA_cleaned_data.csv",
    index_col=0,
    parse_dates=True
)

data.index.name = "Date"

print("Dataset Shape:", data.shape)
display(data.head())
```

```python
duplicate_rows = data.duplicated().sum()
duplicate_dates = data.index.duplicated().sum()

print("Duplicate Records:", duplicate_rows)
print("Duplicate Dates:", duplicate_dates)
```

```python
duplicates = data[data.index.duplicated(keep=False)]

print("Duplicate records found:")
display(duplicates.sort_index())
```

```python
conflicting_dates = data[
    data.index.duplicated(keep=False)
].sort_index()

conflicts = (
    conflicting_dates.groupby(level=0)
    .nunique()
)

conflicts = conflicts[(conflicts > 1).any(axis=1)]

print("Dates with conflicting values:")
display(conflicts)
```

```python
conflicting_dates = data[
    data.index.duplicated(keep=False)
].sort_index()

conflicts = (
    conflicting_dates.groupby(level=0)
    .nunique()
)

conflicts = conflicts[(conflicts > 1).any(axis=1)]

print("Dates with conflicting values:")
display(conflicts)
```

```python
clean_data = data.copy()

clean_data = clean_data[
    ~clean_data.index.duplicated(keep="first")
]

clean_data = clean_data.sort_index()

print("Original Records:", len(data))
print("Records After Cleaning:", len(clean_data))
print("Removed Records:", len(data) - len(clean_data))
```

```python
print("Remaining Duplicate Rows:", clean_data.duplicated().sum())

print("Remaining Duplicate Dates:", clean_data.index.duplicated().sum())

if clean_data.index.is_monotonic_increasing:
    print("Dates are sorted correctly.")

if clean_data.index.is_unique:
    print("All dates are unique.")
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Duplicate Rows",
        "Duplicate Dates"
    ],
    "Before Cleaning": [
        len(data),
        data.duplicated().sum(),
        data.index.duplicated().sum()
    ],
    "After Cleaning": [
        len(clean_data),
        clean_data.duplicated().sum(),
        clean_data.index.duplicated().sum()
    ]
})

display(comparison)
```

```python
clean_data.to_csv("TSLA_deduplicated_data.csv")

print("Cleaned TSLA dataset saved successfully!")
```
