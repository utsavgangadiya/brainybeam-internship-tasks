```python

import streamlit as st
import joblib
import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime
import hashlib
import hmac

# -------------------------------
# 1. PAGE CONFIGURATION
# -------------------------------

st.set_page_config(
    page_title="Heart Disease Prediction",
    page_icon="❤️",
    layout="centered"
)

st.title("❤️ Heart Disease Prediction")
st.write("ML-based heart disease risk prediction demonstration")

# -------------------------------
# 2. MODEL LOADING
# -------------------------------

BASE_DIR = Path(__file__).resolve().parent.parent

MODEL_PATH = (
    BASE_DIR / "saved_models" /
    "heart_disease_model_v1.joblib"
)

@st.cache_resource
def load_model():
    return joblib.load(MODEL_PATH)

try:
    model = load_model()
    st.success("Model loaded successfully")
except Exception as e:
    st.error(f"Model loading failed: {e}")
    st.stop()

# -------------------------------
# 3. BASIC AUTHENTICATION
# -------------------------------

# Configure these credentials in .streamlit/secrets.toml
# Do not hardcode real passwords in your source code.

if "authenticated" not in st.session_state:
    st.session_state.authenticated = False

if not st.session_state.authenticated:
    st.subheader("Login")

    username = st.text_input("Username")
    password = st.text_input("Password", type="password")

    if st.button("Login"):
        try:
            correct_username = st.secrets["APP_USERNAME"]
            correct_password = st.secrets["APP_PASSWORD"]

            if (
                hmac.compare_digest(username, correct_username)
                and hmac.compare_digest(password, correct_password)
            ):
                st.session_state.authenticated = True
                st.rerun()
            else:
                st.error("Invalid username or password")

        except Exception:
            st.error("Authentication credentials are not configured.")

    st.stop()

if st.button("Logout"):
    st.session_state.authenticated = False
    st.rerun()

# -------------------------------
# 4. INPUT VALIDATION
# -------------------------------

st.subheader("Enter Patient Information")

age = st.number_input("Age", 18, 100, 45)
trestbps = st.number_input("Resting Blood Pressure", 60, 250, 120)
chol = st.number_input("Cholesterol", 50, 600, 200)
thalach = st.number_input("Maximum Heart Rate", 50, 250, 150)
oldpeak = st.number_input("Oldpeak", 0.0, 10.0, 1.0)

# -------------------------------
# 5. PREDICTION
# -------------------------------

if st.button("Predict"):

    input_data = pd.DataFrame(
        [[age, trestbps, chol, thalach, oldpeak]],
        columns=[
            "age", "trestbps", "chol",
            "thalach", "oldpeak"
        ]
    )

    try:
        prediction = model.predict(input_data)[0]

        if hasattr(model, "predict_proba"):
            probabilities = model.predict_proba(input_data)[0]
            classes = list(model.classes_)
            positive_index = classes.index(1)
            probability = probabilities[positive_index] * 100
        else:
            probability = None

        st.subheader("Prediction Result")

        if prediction == 1:
            st.warning("Model predicts a positive heart disease result.")
        else:
            st.success("Model predicts a negative heart disease result.")

        if probability is not None:
            st.metric("Model probability for class 1", f"{probability:.2f}%")

        # -------------------------------
        # 6. BASIC MONITORING
        # -------------------------------

        st.subheader("Prediction Monitoring")

        st.write("Prediction time:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
        st.write("Prediction status: Successful")
        st.write("Model version: heart_disease_model_v1")

    except Exception as e:
        st.error(f"Prediction failed: {e}")

st.caption(
    "Educational demonstration only. "
    "This prediction is not a medical diagnosis."
)
```
