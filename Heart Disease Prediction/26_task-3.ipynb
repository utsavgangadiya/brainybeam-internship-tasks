Train a Logistic Regression model from scratch without using scikit-learn, manually implementing gradient descent for optimization, and evaluate it using accuracy, precision, recall, and F1-score computed without built-in metrics functions.

```python
import pandas as pd
import math

# Load datasets
train_df = pd.read_csv("heart_disease_train.csv")
test_df = pd.read_csv("heart_disease_test.csv")

target_column = "target"

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))
```

```python
# Select numerical features
feature_columns = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

print("Features:")
print(feature_columns)
```

```python
# Prepare training data
X_train = []
y_train = []

for i in range(len(train_df)):

    row = []

    for feature in feature_columns:

        row.append(
            float(train_df[feature].iloc[i])
        )

    X_train.append(row)

    y_train.append(
        float(train_df[target_column].iloc[i])
    )


# Prepare testing data
X_test = []
y_test = []

for i in range(len(test_df)):

    row = []

    for feature in feature_columns:

        row.append(
            float(test_df[feature].iloc[i])
        )

    X_test.append(row)

    y_test.append(
        float(test_df[target_column].iloc[i])
    )


print("Training data prepared.")
print("Testing data prepared.")
```

```python
# Calculate training means and standard deviations
means = []
stds = []

for j in range(len(feature_columns)):

    values = []

    for row in X_train:

        values.append(
            row[j]
        )

    total = 0

    for value in values:
        total += value

    mean = total / len(values)

    means.append(mean)

    squared_total = 0

    for value in values:

        squared_total += (
            value - mean
        ) ** 2

    std = math.sqrt(
        squared_total / len(values)
    )

    if std == 0:
        std = 1

    stds.append(std)


# Standardize training data
X_train_scaled = []

for row in X_train:

    new_row = []

    for j in range(len(row)):

        value = (
            row[j] - means[j]
        ) / stds[j]

        new_row.append(value)

    X_train_scaled.append(
        new_row
    )


# Standardize testing data using
# training mean and standard deviation
X_test_scaled = []

for row in X_test:

    new_row = []

    for j in range(len(row)):

        value = (
            row[j] - means[j]
        ) / stds[j]

        new_row.append(value)

    X_test_scaled.append(
        new_row
    )


print("Features standardized successfully.")
```

```python
# Sigmoid function
def sigmoid(z):

    # Prevent overflow
    if z < -500:
        return 0

    if z > 500:
        return 1

    return 1 / (
        1 + math.exp(-z)
    )
```

```python
# Initialize weights
number_of_features = len(
    feature_columns
)

weights = []

for i in range(number_of_features):
    weights.append(0.0)

bias = 0.0


# Training parameters
learning_rate = 0.01
epochs = 5000

number_of_samples = len(
    X_train_scaled
)


# Gradient Descent
for epoch in range(epochs):

    weight_gradients = []

    for i in range(number_of_features):
        weight_gradients.append(0.0)

    bias_gradient = 0.0

    # Calculate gradients
    for i in range(number_of_samples):

        z = bias

        for j in range(number_of_features):

            z += (
                weights[j]
                *
                X_train_scaled[i][j]
            )

        prediction = sigmoid(z)

        error = (
            prediction
            -
            y_train[i]
        )

        for j in range(number_of_features):

            weight_gradients[j] += (
                error
                *
                X_train_scaled[i][j]
            )

        bias_gradient += error


    # Update weights
    for j in range(number_of_features):

        weights[j] -= (
            learning_rate
            *
            weight_gradients[j]
            /
            number_of_samples
        )


    # Update bias
    bias -= (
        learning_rate
        *
        bias_gradient
        /
        number_of_samples
    )


    # Display progress
    if (
        epoch + 1
    ) % 500 == 0:

        print(
            "Epoch:",
            epoch + 1
        )


print("\nTraining completed!")

print("\nWeights:")

for i in range(number_of_features):

    print(
        feature_columns[i],
        ":",
        round(weights[i], 4)
    )

print(
    "Bias:",
    round(bias, 4)
)
```

```python
# Predict probabilities
def predict_probability(X):

    probabilities = []

    for row in X:

        z = bias

        for j in range(
            len(weights)
        ):

            z += (
                weights[j]
                *
                row[j]
            )

        probability = sigmoid(z)

        probabilities.append(
            probability
        )

    return probabilities


# Convert probabilities to classes
def predict_class(probabilities):

    predictions = []

    for probability in probabilities:

        if probability >= 0.5:

            predictions.append(1)

        else:

            predictions.append(0)

    return predictions


# Test predictions
probabilities = predict_probability(
    X_test_scaled
)

predictions = predict_class(
    probabilities
)

print("Predictions generated!")
```

```python
# Initialize confusion matrix values
TP = 0
TN = 0
FP = 0
FN = 0


for i in range(len(y_test)):

    actual = int(
        y_test[i]
    )

    predicted = predictions[i]

    if actual == 1 and predicted == 1:

        TP += 1

    elif actual == 0 and predicted == 0:

        TN += 1

    elif actual == 0 and predicted == 1:

        FP += 1

    elif actual == 1 and predicted == 0:

        FN += 1


print("Confusion Matrix")
print("=" * 40)

print(
    "True Positive:",
    TP
)

print(
    "True Negative:",
    TN
)

print(
    "False Positive:",
    FP
)

print(
    "False Negative:",
    FN
)
```

```python
# Accuracy
total_predictions = (
    TP + TN + FP + FN
)

if total_predictions > 0:

    accuracy = (
        TP + TN
    ) / total_predictions

else:

    accuracy = 0


# Precision
if (
    TP + FP
) > 0:

    precision = (
        TP
        /
        (TP + FP)
    )

else:

    precision = 0


# Recall
if (
    TP + FN
) > 0:

    recall = (
        TP
        /
        (TP + FN)
    )

else:

    recall = 0


# F1 Score
if (
    precision + recall
) > 0:

    f1_score = (
        2
        *
        precision
        *
        recall
        /
        (
            precision
            +
            recall
        )
    )

else:

    f1_score = 0


print("\nMODEL EVALUATION")
print("=" * 40)

print(
    "Accuracy :",
    round(accuracy, 4)
)

print(
    "Precision:",
    round(precision, 4)
)

print(
    "Recall   :",
    round(recall, 4)
)

print(
    "F1 Score :",
    round(f1_score, 4)
)
```

```python
print("\nFINAL RESULTS")
print("=" * 40)

print(
    "Accuracy :",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Precision:",
    round(precision * 100, 2),
    "%"
)

print(
    "Recall   :",
    round(recall * 100, 2),
    "%"
)

print(
    "F1 Score :",
    round(f1_score * 100, 2),
    "%"
)
```
