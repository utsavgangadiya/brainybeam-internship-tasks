Implement user input handling for prediction by validating inputs dynamically, handling edge cases such as missing or incorrect values, and optimizing input processing for real-time predictions.

```python
import pandas as pd
import numpy as np
import joblib

df = pd.read_csv("heart_disease_train.csv")

features = ["age", "trestbps", "chol", "thalach", "oldpeak"]

print("Required Features:")
print(features)

print("\nDataset Information:")
print(df[features].info())
```

```python
validation_rules = {
    "age": (1, 120),
    "trestbps": (50, 250),
    "chol": (50, 700),
    "thalach": (40, 250),
    "oldpeak": (0, 10)
}
```

```python
def validate_input(user_data):

    errors = {}
    cleaned_data = {}

    for feature in features:

        value = user_data.get(feature)

        if value is None or str(value).strip() == "":
            errors[feature] = "Value is missing"
            continue

        try:
            value = float(value)
        except (ValueError, TypeError):
            errors[feature] = "Enter a valid numeric value"
            continue

        if not np.isfinite(value):
            errors[feature] = "Value must be finite"
            continue

        minimum, maximum = validation_rules[feature]

        if value < minimum or value > maximum:
            errors[feature] = f"Value must be between {minimum} and {maximum}"
            continue

        cleaned_data[feature] = value

    if errors:
        return None, errors

    return pd.DataFrame([cleaned_data], columns=features), {}
```

```python
user_data = {}

for feature in features:
    user_data[feature] = input(f"Enter {feature}: ")

print("\nUser Input:")
print(user_data)
```

```python
validated_input, errors = validate_input(user_data)

if errors:
    print("Input Validation Failed:")

    for feature, error in errors.items():
        print(f"{feature}: {error}")

else:
    print("All inputs are valid.")
    display(validated_input)
```

```python
import os

for file in os.listdir():
    if file.endswith((".pkl", ".joblib")):
        print(file)
        
model = joblib.load("saved_models/heart_disease_model_v1.joblib")

print("Model loaded successfully!")
```

```python
if validated_input is not None:

    prediction = model.predict(validated_input)[0]

    if prediction == 1:
        print("Heart Disease: Positive")
    else:
        print("Heart Disease: Negative")

else:
    print("Prediction cancelled due to invalid inputs.")
```

```python
test_cases = [
    {"age": "45", "trestbps": "120", "chol": "200", "thalach": "150", "oldpeak": "1.2"},
    {"age": "", "trestbps": "120", "chol": "200", "thalach": "150", "oldpeak": "1.2"},
    {"age": "abc", "trestbps": "120", "chol": "200", "thalach": "150", "oldpeak": "1.2"},
    {"age": "150", "trestbps": "120", "chol": "200", "thalach": "150", "oldpeak": "1.2"}
]

for i, test in enumerate(test_cases, 1):
    result, errors = validate_input(test)

    print(f"\nTest Case {i}:")

    if errors:
        print("Invalid:", errors)
    else:
        print("Valid input")
        print(result)
```

```python
import time

start = time.perf_counter()

validated_input, errors = validate_input(user_data)

if validated_input is not None:
    prediction = model.predict(validated_input)[0]
    print("Prediction:", prediction)
else:
    print("Invalid input:", errors)

end = time.perf_counter()

print("Processing Time:", round((end - start) * 1000, 3), "ms")
```
