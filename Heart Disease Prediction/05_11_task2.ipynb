# Heart Disease Prediction – Final Project Documentation

## 1. Project Overview

The Heart Disease Prediction project uses Machine Learning to predict the possibility of heart disease based on patient health information.

The project includes model development, prediction functionality, performance evaluation, latency optimization, and a Streamlit-based user interface.

**Project Objectives:**
- Develop a Machine Learning model for heart disease prediction.
- Process patient information and generate predictions.
- Evaluate model performance.
- Measure prediction latency.
- Implement caching and model compression.
- Build a user-friendly application using Streamlit.

## 3. Implementation Methodology

### 3.1 Data Preparation

The dataset was used to prepare patient information for Machine Learning predictions.

The model uses five input features:

- Age
- Resting blood pressure (trestbps)
- Cholesterol (chol)
- Maximum heart rate (thalach)
- ST depression (oldpeak)

Input features were arranged in the same order used during model training.

### 3.2 Model Integration

The trained Machine Learning model was saved using Joblib and loaded into the prediction pipeline.

The model generates predictions using the `predict()` function.

### 3.3 Prediction Pipeline

The prediction process consists of:

1. Accepting patient input.
2. Preparing input data.
3. Converting data into the required numerical format.
4. Passing input to the trained model.
5. Generating the prediction.
6. Measuring prediction execution time.

### 3.4 Application Development

A Streamlit application was developed to provide an interface for entering patient information and displaying prediction results.

Basic authentication and input validation were also considered in the application implementation.

## 4. Experimental Work

### 4.1 Prediction Latency Testing

Python's `time.perf_counter()` was used to measure execution time.

The following measurements were collected:

- Data-processing time
- Model inference time
- Total prediction latency
- Average latency
- Minimum latency
- Maximum latency
- 95th-percentile latency

A benchmark of 1,000 prediction executions was implemented.

### 4.2 Caching Optimization

LRU caching was implemented using Python's `functools` library.

The purpose was to avoid repeating model inference for identical input values.

The first prediction requires model inference, while repeated identical inputs can retrieve the stored result.

### 4.3 Model Compression

Joblib compression was tested to compare the original model size with the compressed model size.

Compression was evaluated as a storage optimization technique. It was not considered model quantization.


**Result Analysis:** The experiments were designed to compare prediction execution time, repeated prediction performance, and model storage size. Actual measured values should be added after running the notebook.

## 5. Performance Optimization

The following optimization techniques were implemented or evaluated:

1. **Model Loading:** Load the trained model once rather than repeatedly initializing it.
2. **Data Processing:** Convert prediction inputs into the numerical format expected by the model.
3. **Prediction Caching:** Store results for repeated identical inputs.
4. **Model Compression:** Reduce model storage size using Joblib compression.
5. **Latency Profiling:** Measure different stages of the prediction pipeline.

## 6. Deployment Challenges

During deployment and application development, the following challenges were encountered:

- Dependency and package configuration issues.
- Model file path management.
- Scikit-learn feature-name warnings caused by inconsistent input formats.
- GitHub synchronization and push conflicts.
- Streamlit Community Cloud dependency resolution issues.

The feature-name warning was addressed by providing NumPy arrays in the same feature order used during model training.

The Streamlit deployment encountered a requirements-file path resolution issue. Successful deployment should only be claimed after the application runs correctly in the cloud.

## 7. Limitations

- Local inference benchmarks do not measure actual network API latency.
- Sequential prediction testing does not demonstrate concurrent request capacity.
- Caching benefits repeated inputs, not unique predictions.
- Joblib compression does not guarantee faster inference.
- Model quantization was not implemented.
- Clinical accuracy and reliability require further validation.
